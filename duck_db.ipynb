```python
import duckdb
```

```python
con = duckdb.connect("nba.duckdb")

con.execute("""
            CREATE OR REPLACE TABLE games AS
            SELECT * FROM read_csv_auto('data/games.csv')
            """)

con.execute("""
            CREATE OR REPLACE TABLE player_boxscores AS
            SELECT * FROM read_csv_auto('data/player_boxscores.csv')
            """)
con.execute("""
            CREATE OR REPLACE TABLE players AS
            SELECT * FROM read_csv_auto('data/players.csv')
            """)
```

```python
for table in ['games', 'players', 'player_boxscores']:
    count = con.execute(f"SELECT COUNT(*) FROM {table}").fetchone()[0]
    print(f"{table}: {count} rows")

con.close()
```
